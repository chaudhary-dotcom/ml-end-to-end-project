```python
def get_product(x: int, y:int) -> int:
    return x*y
```

```python
get_product(5,5)
```

```python
get_product(5, '2')
```

```python
from ensure import ensure_annotations
```

```python
@ensure_annotations
def get_multiply(x: int, y:int) -> int:
    return x*y

```

```python
get_multiply(2,2)
```

```python
get_multiply(2, '2')
```
